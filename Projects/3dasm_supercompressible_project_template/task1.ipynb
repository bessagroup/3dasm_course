```python
from pathlib import Path
import pickle

RUN_AGENTS = False  # True starts the adda campaigns again; keep False in the notebook you submit
RECOMPUTE = False   # True recomputes every cached result from the data

DATA = Path("data")  # supercompressible_3d/ and supercompressible_7d/, as released
CACHE = Path("cache") / "task1"
CACHE.mkdir(parents=True, exist_ok=True)


def cached(name, compute):
    """Return compute(), stored in CACHE/<name>.pkl; it is computed again only if RECOMPUTE is True or the file is missing."""
    file = CACHE / f"{name}.pkl"
    if RECOMPUTE or not file.exists():
        result = compute()
        file.write_bytes(pickle.dumps(result))
        return result
    return pickle.loads(file.read_bytes())
```
